# 7. Route C — Die Kontraktions-Eichung

## Worum es geht

Route B aus den Kapiteln 5 und 6 funktioniert, hat aber zwei Löcher, die beide *dieselbe* Ursache haben:

* Der Companion-Operator $E$ schneidet die Gedächtnissumme nach $K$ Termen ab. Alles, was länger als $K\Delta t$ zurückliegt, ist verloren (der Rest-Term $R_n$ aus der Tabelle in dem Abschnitt *Der Companionpropagator*).
* Das Historien-Fenster $X_0 = (\rho_{K-1},\dots,\rho_0)^{\mathsf T}$ muss gefüllt werden, bevor das Gitter überhaupt starten darf. Diese $K$ Zustände rechnet der klassische Computer. Bei $n=100$ Schritten und $K=25$ sind das bereits $25\,\%$ der gesamten Dynamik.

Route C schließt beide Löcher, und zwar nicht durch einen besseren Gedächtniskern, sondern durch das Gegenteil: **wir werfen den Gedächtniskern ganz weg.** Der Trick besteht darin, eine Ebene *höher* anzusetzen — nicht bei den reduzierten Karten $\mathcal{E}_n$, sondern direkt bei dem Objekt, aus dem wir sie im Abschnitt *Konstruktion der dynamischen Karten* $\mathcal{E}_n$ *via HEOM* überhaupt erst gewonnen haben: dem HEOM-Propagator

$$P = \exp(\mathcal{L}_{\mathrm{HEOM}}\,\Delta t) \in \mathbb{C}^{\mathcal{D}_{\mathrm{tot}}\times\mathcal{D}_{\mathrm{tot}}},\qquad \mathcal{D}_{\mathrm{tot}} = \mathcal{N}_{\mathrm{ADOs}}\cdot d^2 .$$

Dieses $P$ haben wir längst berechnet. Wir haben es bisher nur benutzt, um daraus die Kurzzeit-Karten $\mathcal{E}_n$ zu erzeugen, und es danach weggeworfen. Abschnitt 7.1 zeigt, dass genau in diesem Wegwerfen der ganze Fehler steckt.

Der Preis dafür ist ein neues Hindernis, das mit dem Gedächtnis nichts mehr zu tun hat: $P$ hat zwar den Spektralradius $\rho(P)=1$, aber die Operatornorm $\Vert P\Vert_2 = 32{,}39$. Nach Proposition 13 multiplizieren sich die Erfolgswahrscheinlichkeiten, und $32{,}39^{-200}$ ist ungefähr $10^{-301}$. Abschnitt 7.2 zeigt, dass dieses Hindernis fundamental aussieht, und Abschnitt 7.3 zeigt, dass es das nicht ist: Die Zahl $32{,}39$ ist eine Aussage über das *Skalarprodukt*, nicht über die Physik. Man kann das Skalarprodukt wechseln, und danach ist $P$ eine Kontraktion.

Die Reihenfolge ist dabei zwingend. Erst die Eichung, dann die Arnoldi-Kompression — in Abschnitt 7.4 sehen wir, dass die Kompression in der neuen Eichung nicht nur funktioniert, sondern dass der unangenehme Trade-off aus Kapitel 6 (zu großes $m$ lässt $\Vert H_m\Vert_2$ wieder ansteigen) vollständig verschwindet.

## 7.1 Der Gedächtnis-Rest $R_n$ ist selbstgemacht

Erinnern wir uns an das Ergebnis aus dem Abschnitt *Konstruktion der dynamischen Karten* $\mathcal{E}_n$ *via HEOM*. Dort haben wir die dynamischen Karten aus dem Propagator gewonnen:

$$\mathcal{E}_n = Q\,P^n\,Q^{\mathsf T},\qquad Q = \begin{pmatrix}\mathbb{1}_{d^2} & \mathbf{0}\end{pmatrix},\qquad Q^{\mathsf T} = \begin{pmatrix}\mathbb{1}_{d^2}\\ \mathbf{0}\end{pmatrix} \tag{7.1}$$

Diese Gleichung war **exakt** — wir haben sie hergeleitet, ohne irgendetwas abzuschneiden. Und die darunterliegende Gleichung war noch einfacher:

$$\vec{\Sigma}(n\Delta t) = P^n\,\vec{\Sigma}(0),\qquad \vec{\Sigma}(0) = \begin{pmatrix}\mathrm{vec}(\rho_S(0))\\ \mathbf{0}\end{pmatrix} \tag{7.2}$$

Halten wir kurz inne. Gleichung (7.2) sagt: um $\rho_S$ zu einem beliebigen Zeitpunkt $n\Delta t$ zu bekommen, wende ich $n$-mal dieselbe Matrix an und lese den obersten Block ab. Kein $K$, keine Transfer-Tensoren, keine abgeschnittene Summe, kein klassisch vorberechnetes Fenster. Der Grund ist, dass die ADOs das Gedächtnis bereits *tragen*: Sie sind genau dafür da.

> **Proposition 18.** Auf dem erweiterten Raum $\mathbb{C}^{\mathcal{D}_{\mathrm{tot}}}$ ist die HEOM-Zeitentwicklung eine Halbgruppe, d.h. $\vec{\Sigma}_n = P\,\vec{\Sigma}_{n-1}$ hängt nur vom unmittelbaren Vorgänger ab. Die gesamte Nicht-Markovianität von $\rho_S$ steckt allein darin, dass der Projektor $Q$ nicht invertierbar ist. Insbesondere verschwindet der Gedächtnis-Rest $R_n$ aus dem Abschnitt *Der Companionpropagator* für **jedes** $n$ identisch, ohne dass ein $K$ gewählt werden muss.

**Beweis.** Der HEOM-Liouvillian $\mathcal{L}_{\mathrm{HEOM}}$ ist zeitunabhängig, also gilt für Matrixexponentiale mit demselben Erzeuger $e^{\mathcal{L}(t+s)} = e^{\mathcal{L}t}e^{\mathcal{L}s}$ und damit

$$\vec{\Sigma}\big((n)\Delta t\big) = e^{\mathcal{L}n\Delta t}\vec{\Sigma}(0) = e^{\mathcal{L}\Delta t}\,e^{\mathcal{L}(n-1)\Delta t}\vec{\Sigma}(0) = P\,\vec{\Sigma}\big((n-1)\Delta t\big).$$

Das ist eine Ein-Schritt-Rekursion: Zur Berechnung von $\vec{\Sigma}_n$ wird ausschließlich $\vec{\Sigma}_{n-1}$ gebraucht, keine älteren Zustände. Ein Gedächtnis-Rest kann also gar nicht entstehen.

Warum ist $\rho_S$ dann trotzdem nicht-Markovsch? Weil die Halbgruppen-Eigenschaft beim Projizieren kaputtgeht. Setzen wir (7.1) zusammen:

$$\mathcal{E}_{n}\,\mathcal{E}_{m} = \big(Q P^{n} Q^{\mathsf T}\big)\big(Q P^{m} Q^{\mathsf T}\big) = Q P^{n}\,\underbrace{\big(Q^{\mathsf T} Q\big)}_{\neq\,\mathbb{1}}\,P^{m} Q^{\mathsf T} \;\neq\; Q P^{n+m} Q^{\mathsf T} = \mathcal{E}_{n+m}$$

Der Störenfried in der Mitte ist explizit

$$Q^{\mathsf T}Q = \begin{pmatrix}\mathbb{1}_{d^2}\\ \mathbf{0}\end{pmatrix}\begin{pmatrix}\mathbb{1}_{d^2} & \mathbf{0}\end{pmatrix} = \begin{pmatrix}\mathbb{1}_{d^2} & \mathbf{0}\\ \mathbf{0} & \mathbf{0}\end{pmatrix},$$

also der Operator, der **alle ADOs auf null setzt**. Wer zwischendurch auf den Systemblock projiziert, löscht das Bad-Gedächtnis; und genau dieses gelöschte Gedächtnis muss die Transfer-Tensor-Reihe $\sum_m T_m\rho_{n-m}$ hinterher mühsam wieder rekonstruieren. Vermeidet man die Projektion, gibt es nichts zu rekonstruieren. $\;\blacksquare$

**Dieselbe Tabelle wie beim Companionpropagator, aber mit dem ADO-Register:**

| Schritt | Propagation auf dem ADO-Register | Exakter Zustand | Fehler $R_n$ |
| :---: | :--- | :--- | :---: |
| **$n=1$** | $\vec{\Sigma}_1 = P\begin{pmatrix}\mathrm{vec}\rho_0\\ \mathbf{0}\end{pmatrix}$ | $\vec{\Sigma}_1$ | $0$ |
| **$n=2$** | $\vec{\Sigma}_2 = P\,\vec{\Sigma}_1 = P^2\vec{\Sigma}_0$ | $\vec{\Sigma}_2$ | $0$ |
| **$n=3$** | $\vec{\Sigma}_3 = P\,\vec{\Sigma}_2 = P^3\vec{\Sigma}_0$ | $\vec{\Sigma}_3$ | $0$ |
| **$n=4$** | $\vec{\Sigma}_4 = P\,\vec{\Sigma}_3 = P^4\vec{\Sigma}_0$ | $\vec{\Sigma}_4$ | $\mathbf{0}$ |
| $\vdots$ | $\vdots$ | $\vdots$ | $\vdots$ |
| **$n=100$** | $\vec{\Sigma}_{100} = P\,\vec{\Sigma}_{99} = P^{100}\vec{\Sigma}_0$ | $\vec{\Sigma}_{100}$ | $\mathbf{0}$ |

Die Spalte, die dort ab $n=K+1$ mit $T_{K+1}\rho_0,\;T_{K+2}\rho_0,\dots$ zu wuchern begann, bleibt hier für immer leer.

**Der Preis.** Umsonst ist das nicht: Das Register wird größer. Für unser 4-Site-FMO-Modell mit Hierarchietiefe 3 und $\mathcal{N}_{\mathrm{ADOs}} = 165$ ist

$$\mathcal{D}_{\mathrm{tot}} = 165\cdot 4^2 = 2640 \qquad\text{gegenüber}\qquad K\cdot d^2 = 25\cdot 16 = 400$$

für das Historien-Fenster. Auf den ersten Blick haben wir also getauscht: exakte Dynamik gegen ein $6{,}6$-mal größeres Register. Abschnitt 7.4 zeigt, dass dieser Tausch fast gratis ist — nach der Arnoldi-Kompression landen wir bei 7 bis 8 Qubits gegenüber den 6 von Route B, obwohl das unkomprimierte Register $6{,}6$-mal größer ist.

**Und die Rechenzeit?** Numerisch gemessen (4-Site-FMO, $\lambda=3$, $\gamma=50\,\mathrm{cm}^{-1}$, $T=300\,\mathrm{K}$, $\Delta t = 10\,\mathrm{fs}$, 100 Schritte, jeweils gegen den `HEOMSolver` von QuTiP):

| Verfahren | klassisch propagierte Zeitschritte | Abweichung von QuTiP |
| :--- | :---: | :---: |
| Companion, $K=10$ | 10 | $5{,}66\times10^{-2}$ |
| Companion, $K=25$ | 25 | $5{,}58\times10^{-3}$ |
| Companion, $K=40$ | 40 | $9{,}03\times10^{-4}$ |
| **$P$ direkt iterieren** | **0** | $\mathbf{5{,}27\times10^{-7}}$ |

Die $5{,}27\times10^{-7}$ sind übrigens nicht *unser* Fehler, sondern die Integrationstoleranz des QuTiP-Solvers selbst — wir vergleichen hier zwei Verfahren, die dieselbe exakte Gleichung lösen, und sie stimmen bis auf die Genauigkeit des Referenzverfahrens überein.

## 7.2 Das neue Hindernis: $\rho(P) = 1$, aber $\Vert P\Vert_2 = 32{,}39$

Proposition 13 aus Kapitel 5 gilt wörtlich weiter, wir müssen nur $E$ durch $P$ und $X_0$ durch $\vec{\Sigma}_0$ ersetzen. Der Beweis benutzte ausschließlich das Teleskopprodukt und nichts, was für $E$ spezifisch gewesen wäre:

$$p_{\mathrm{total}} = \underbrace{\frac{1}{s^{2n}}}_{\text{Teil 1 (Gatter-Strafe)}}\cdot\underbrace{\frac{\Vert P^n\vec{\Sigma}_0\Vert^2}{\Vert\vec{\Sigma}_0\Vert^2}}_{\text{Teil 2 (physikalischer Zerfall)}},\qquad s = \Vert P\Vert_2 \tag{7.3}$$

Bei Route B war Teil 2 harmlos und Teil 1 die Katastrophe. Hier ist die Lage anders — und lehrreicher: **beide Faktoren sind sinnlos**, in entgegengesetzte Richtungen.

**Teil 1.** Für unser Modell messen wir

$$\rho(P) = 1{,}000000 \qquad\text{aber}\qquad \Vert P\Vert_2 = 32{,}39 \quad\implies\quad \text{Teil 1} = \frac{1}{32{,}39^{200}} = 8{,}17\times10^{-303}$$

**Teil 2.** Man würde nun erwarten, dass der physikalische Zerfall diesen Faktor bestenfalls unverändert lässt. Tatsächlich *wächst* die euklidische Länge des ADO-Vektors entlang der echten Trajektorie ganz erheblich:

| $n$ | $0$ | $1$ | $2$ | $5$ | $10$ | $25$ | $50$ | $100$ |
| :--- | ---: | ---: | ---: | ---: | ---: | ---: | ---: | ---: |
| $\Vert P^n\vec\Sigma_0\Vert/\Vert\vec\Sigma_0\Vert$ | $1$ | $1{,}98$ | $19{,}8$ | $432$ | $1872$ | $2950$ | $2717$ | $1101$ |

Das Maximum liegt bei $3165$, und damit ist $\text{Teil 2} = 1101^2 = 1{,}21\times10^{6}$. Zusammen ergibt sich genau der gemessene Wert

$$p_{\mathrm{total}}(100) = \underbrace{8{,}17\times10^{-303}}_{\text{Teil 1}}\cdot\underbrace{1{,}21\times10^{6}}_{\text{Teil 2}} = 9{,}90\times10^{-297}$$

**Warum wächst der Vektor?** Das ist kein Widerspruch zur Physik, sondern der Hinweis darauf, dass die euklidische Norm auf dem ADO-Raum der falsche Maßstab ist. Von der Länge $\Vert\vec\Sigma_{100}\Vert$ entfallen nämlich nur $5{,}2\times10^{-4}$ auf den Systemblock $\mathrm{vec}\,\rho_S$; die restlichen $99{,}95\,\%$ stecken in den ADOs. Und die ADOs *sind keine Dichtematrizen* — sie sind Buchhaltungsgrößen für die Bad-Korrelationen und tragen in der HEOM-Konvention große numerische Vorfaktoren. Es gibt keinerlei physikalischen Grund, warum sie beschränkt bleiben sollten. Der physikalische Anteil verhält sich dagegen tadellos:

$$\Vert\mathrm{vec}\,\rho_S(1\,\mathrm{ps})\Vert = 0{,}573 \qquad\text{bei}\qquad \mathrm{Tr}\,\rho_S(1\,\mathrm{ps}) = 1{,}000000$$

Damit steht das Problem in seiner eigentlichen Form da. Wir messen Längen mit einem Maßstab, der die interessante Größe zu $0{,}05\,\%$ und die Buchhaltung zu $99{,}95\,\%$ gewichtet. Beide Faktoren in (7.3) sind deshalb astronomisch, sie heben sich fast auf — aber eben nur fast, und was übrig bleibt, ist um $278$ Größenordnungen schlechter als die $6{,}7\times10^{-24}$ von Route B. Abschnitt 7.3 wechselt den Maßstab; danach ist jeder der beiden Faktoren für sich von der Ordnung $1$.

**Wie kann der Spektralradius 1 und die Norm 32 sein?** Weil $P$ stark **nicht-normal** ist ($PP^\dagger \neq P^\dagger P$). Bei einer normalen Matrix wären Spektralradius und Operatornorm identisch; bei einer nicht-normalen können sie beliebig weit auseinanderliegen. Ein $2\times2$-Beispiel macht das sofort klar:

$$M = \begin{pmatrix}0{,}9 & 30\\ 0 & 0{,}5\end{pmatrix} \quad\implies\quad \underbrace{\rho(M) = 0{,}9}_{\text{Eigenwerte } 0{,}9,\;0{,}5} \qquad\text{und}\qquad \underbrace{\Vert M\Vert_2 = 30{,}02}_{\text{größter Singulärwert}}$$

Jede Trajektorie $M^n x$ fällt hier exponentiell gegen null — der Spektralradius sagt, was auf lange Sicht passiert. Die Operatornorm sagt dagegen, was im *schlimmsten Einzelschritt* passieren kann, und dieser schlimmste Fall wird von dem einen Testvektor $x=(0,1)^{\mathsf T}$ ausgelöst, auf dem $Mx = (30,\,0{,}5)^{\mathsf T}$ tatsächlich um den Faktor $30$ wächst. Die Sz.-Nagy-Dilatation kennt aber nur die Operatornorm: Sie zwingt uns $s=30{,}02$ auf, egal wie brav die tatsächliche Trajektorie ist. Genau das ist unsere Lage mit $P$, nur in $2640$ Dimensionen statt in zweien.

**Kann man das durch feinere Zeitschritte reparieren?** Die Hoffnung wäre: kleineres $\Delta t$ bedeutet $P$ näher an $\mathbb{1}$, also $s$ näher an $1$. Das stimmt auch — aber man braucht dann eben mehr Schritte, und der Exponent $n$ wächst genauso schnell, wie $\ln s$ fällt. Präzise:

> **Proposition 19.** Sei die Gesamtzeit $T = n\Delta\tau$ fest und $s(\Delta\tau) = \Vert e^{\mathcal{L}\Delta\tau}\Vert_2$. Dann gilt
> $$\lim_{\Delta\tau\to 0}\; n\,\ln s(\Delta\tau) \;=\; T\,\omega(\mathcal{L}),\qquad \omega(\mathcal{L}) := \lambda_{\max}\!\left(\frac{\mathcal{L}+\mathcal{L}^\dagger}{2}\right) \tag{7.4}$$
> Die Gatter-Strafe $s^{-2n} = e^{-2n\ln s}$ läuft für $\Delta\tau\to0$ also **nicht** gegen $1$, sondern gegen die endliche, von der Diskretisierung unabhängige Zahl $e^{-2T\omega(\mathcal{L})}$. Die Größe $\omega(\mathcal{L})$ heißt numerische Abszisse.

**Beweis.** Wir differenzieren die Länge einer Trajektorie $x(\tau) = e^{\mathcal{L}\tau}x_0$ bei $\tau=0$. Mit $\dot{x} = \mathcal{L}x$ gilt

$$\begin{aligned}
\frac{\mathrm d}{\mathrm d\tau}\Vert x(\tau)\Vert^2
&= \frac{\mathrm d}{\mathrm d\tau}\big(x^\dagger x\big)
= \dot{x}^\dagger x + x^\dagger \dot{x}
= (\mathcal{L}x)^\dagger x + x^\dagger(\mathcal{L}x)
= x^\dagger\big(\mathcal{L}^\dagger + \mathcal{L}\big)x
= 2\,x^\dagger\left(\frac{\mathcal{L}+\mathcal{L}^\dagger}{2}\right)x \;\le\; 2\,\omega(\mathcal{L})\,\Vert x\Vert^2
\end{aligned}$$

wobei die Ungleichung die Rayleigh-Abschätzung für die *hermitesche* Matrix $\tfrac12(\mathcal{L}+\mathcal{L}^\dagger)$ ist, deren größter Eigenwert per Definition $\omega(\mathcal{L})$ ist. Gleichheit tritt genau dann ein, wenn $x$ der zugehörige Eigenvektor ist — wählt man diesen als $x_0$, wird die Abschätzung scharf. Also gilt für kleine $\Delta\tau$

$$\Vert e^{\mathcal{L}\Delta\tau}\Vert_2 = \max_{\Vert x_0\Vert=1}\Vert x(\Delta\tau)\Vert = 1 + \omega(\mathcal{L})\,\Delta\tau + \mathcal{O}(\Delta\tau^2)$$

und damit, mit $\ln(1+u) = u + \mathcal{O}(u^2)$ und $n = T/\Delta\tau$,

$$n\,\ln s(\Delta\tau) = \frac{T}{\Delta\tau}\Big(\omega(\mathcal{L})\,\Delta\tau + \mathcal{O}(\Delta\tau^2)\Big) = T\,\omega(\mathcal{L}) + \mathcal{O}(\Delta\tau) \;\xrightarrow[\Delta\tau\to0]{}\; T\,\omega(\mathcal{L}) \qquad\blacksquare$$

Für unser Modell messen wir $\omega(\mathcal{L}) = +2{,}61\times10^{3}\,\mathrm{cm}^{-1}$, und die Gesamtzeit $T=1000\,\mathrm{fs}$ entspricht $T = 0{,}1884\,\mathrm{cm}$ in den Einheiten $\tau = 2\pi c\,t$. Damit ist

$$T\,\omega(\mathcal{L}) = 0{,}1884 \cdot 2{,}61\times10^{3} \approx 4{,}9\times10^{2} \quad\implies\quad p_{\mathrm{total}} \;\gtrsim\; e^{-2\cdot 492} = e^{-984} \approx 10^{-427}$$

Es hilft also überhaupt nichts, feiner zu diskretisieren. Im Gegenteil zeigt die Messung, dass *gröbere* Schritte etwas besser sind (bei $\Delta t = 10\,\mathrm{fs}$ ist $n\ln s = 348$ statt der $487$ bei $1\,\mathrm{fs}$), aber man kann die Schritte nur so grob machen, wie die Dynamik es erlaubt.

**Was sonst noch nicht funktioniert.** Drei naheliegende Reparaturversuche wurden numerisch durchgemessen und verworfen, bevor Abschnitt 7.3 entstand:

| Versuch | Idee | Gemessenes Ergebnis |
| :--- | :--- | :--- |
| Arnoldi direkt auf $P$ | wie Kapitel 6: auf die physikalische Trajektorie einschränken | $\Vert H_m\Vert \approx 32$ bleibt; Trajektorie divergiert auf $10^{65}$ |
| Diagonales Balancing von $\mathcal{L}$ | die ADO-Skalen ausgleichen (`scipy.linalg.matrix_balance`) | $\Vert P\Vert:\;32{,}4\to2{,}54$; immer noch tödlich, $2{,}54^{200} = 10^{81}$ |
| Trotter-Aufspaltung $\mathcal{L} = \mathcal{L}_{\mathrm{ah}} + \mathcal{L}_{\mathrm{h}}$ | unitären und dissipativen Teil trennen | $\lambda_{\max}(\mathcal{L}_{\mathrm{h}}) = +2613 > 0$, der „dissipative" Faktor ist gar keine Kontraktion |

Die erste Zeile ist die wichtigste: Arnoldi hat im Abschnitt *Arnoldi compression* funktioniert, weil dort $\Vert E\Vert_2 = \sqrt{2}$ nur knapp über 1 lag und die physikalische Trajektorie diesen Worst-Case gar nicht erst besuchte. Bei $\Vert P\Vert_2 = 32$ liegen Worst-Case und Trajektorie so weit auseinander, dass die Kompression numerisch instabil wird. Man braucht erst eine bessere Basis, dann die Kompression.

## 7.3 Die Eichung: ein Skalarprodukt, in dem $\mathcal{L}$ dissipativ ist

Kehren wir zum $2\times2$-Beispiel zurück, denn dort steckt die ganze Lösung schon drin. Wir hatten

$$M = \begin{pmatrix}0{,}9 & 30\\ 0 & 0{,}5\end{pmatrix},\qquad \rho(M) = 0{,}9,\qquad \Vert M\Vert_2 = 30{,}02 .$$

Die $30$ in der Nebendiagonale ist der Übeltäter. Aber sie ist gar keine Eigenschaft der Abbildung, sondern eine Eigenschaft der *Koordinaten*, in denen wir sie aufschreiben. Skalieren wir die zweite Koordinate um einen Faktor $c$, also $\tilde{x} = W^{1/2}x$ mit $W^{1/2} = \mathrm{diag}(1,c)$, so lautet dieselbe Abbildung in den neuen Koordinaten

$$\tilde M = W^{1/2}\,M\,W^{-1/2} = \begin{pmatrix}1 & 0\\ 0 & c\end{pmatrix}\begin{pmatrix}0{,}9 & 30\\ 0 & 0{,}5\end{pmatrix}\begin{pmatrix}1 & 0\\ 0 & 1/c\end{pmatrix} = \begin{pmatrix}0{,}9 & 30/c\\ 0 & 0{,}5\end{pmatrix}$$

Mit $c=300$ wird daraus $\tilde M = \left(\begin{smallmatrix}0{,}9 & 0{,}1\\ 0 & 0{,}5\end{smallmatrix}\right)$ und damit

$$\Vert \tilde M\Vert_2 = 0{,}908 \;<\; 1,\qquad \rho(\tilde M) = 0{,}9\;\text{ (unverändert)}$$

Aus einer Matrix mit Norm $30$ ist eine **Kontraktion** geworden, ohne dass sich an der Dynamik das Geringste geändert hätte: $\tilde M$ ist zu $M$ ähnlich, hat also dieselben Eigenwerte und beschreibt dieselbe Trajektorie, nur in anderen Koordinaten aufgeschrieben. Für die Sz.-Nagy-Dilatation ist der Unterschied aber alles: statt $s = 30{,}02$ genügt jetzt $s = 1$.

Genau das machen wir mit $P$. Nur können wir die richtigen Koordinaten in $2640$ Dimensionen nicht raten — wir müssen sie ausrechnen. Das Werkzeug dafür ist die Lyapunov-Gleichung.

**Das $W$-Skalarprodukt.** Sei $W$ hermitesch und positiv definit. Wir definieren

$$\langle x,y\rangle_W := x^\dagger W y,\qquad \Vert x\Vert_W := \sqrt{\langle x,x\rangle_W} = \Vert W^{1/2}x\Vert_2,\qquad \Vert A\Vert_W := \sup_{x\neq0}\frac{\Vert Ax\Vert_W}{\Vert x\Vert_W}$$

$W^{1/2}$ ist dabei die eindeutige positiv definite Wurzel, die man über die Spektralzerlegung $W = V\,\mathrm{diag}(w_i)\,V^\dagger$ als $W^{1/2} = V\,\mathrm{diag}(\sqrt{w_i})\,V^\dagger$ bekommt.

> **Proposition 20 (Lyapunov-Eichung).** Sei $\delta>0$ so gewählt, dass $\mathcal{L}-\delta\mathbb{1}$ Hurwitz ist, d.h. $\mathrm{Re}\,\lambda < \delta$ für jeden Eigenwert $\lambda$ von $\mathcal{L}$. Dann besitzt die Lyapunov-Gleichung
> $$(\mathcal{L}-\delta\mathbb{1})^\dagger W + W(\mathcal{L}-\delta\mathbb{1}) = -\,\mathbb{1} \tag{7.5}$$
> genau eine Lösung $W$; diese ist hermitesch und positiv definit, und es gilt
> $$\Vert e^{\mathcal{L}\tau}\Vert_W \;\le\; e^{\delta\tau}\qquad\text{für alle }\tau\ge0 . \tag{7.6}$$

**Beweis.** Wir zeigen drei Dinge nacheinander.

**(a) Der Kandidat existiert und ist positiv definit.** Weil $\delta\mathbb{1}$ mit $\mathcal{L}$ kommutiert, gilt $e^{(\mathcal{L}-\delta)\tau} = e^{-\delta\tau}e^{\mathcal{L}\tau}$. Wir setzen

$$W := \int_0^\infty \Big(e^{(\mathcal{L}-\delta)\tau}\Big)^\dagger\,e^{(\mathcal{L}-\delta)\tau}\,\mathrm d\tau = \int_0^\infty e^{-2\delta\tau}\,e^{\mathcal{L}^\dagger\tau}\,e^{\mathcal{L}\tau}\,\mathrm d\tau \tag{7.7}$$

Da $\mathcal{L}-\delta$ Hurwitz ist, existieren $c,\eta>0$ mit $\Vert e^{(\mathcal{L}-\delta)\tau}\Vert_2 \le c\,e^{-\eta\tau}$, und der Integrand ist in der Norm durch $c^2e^{-2\eta\tau}$ beschränkt — das Integral konvergiert absolut. Hermitezität ist offensichtlich, denn der Integrand hat die Form $Z^\dagger Z$. Positive Definitheit ebenso, denn für $x\neq0$ ist

$$x^\dagger W x = \int_0^\infty \big\Vert e^{(\mathcal{L}-\delta)\tau}x\big\Vert^2\,\mathrm d\tau \;>\;0,$$

weil der Integrand stetig, nichtnegativ und bei $\tau=0$ gleich $\Vert x\Vert^2>0$ ist.

**(b) Der Kandidat löst (7.5).** Wir setzen $G(\tau) := \big(e^{(\mathcal{L}-\delta)\tau}\big)^\dagger e^{(\mathcal{L}-\delta)\tau}$ und differenzieren. Weil $(\mathcal{L}-\delta)^\dagger$ mit $e^{(\mathcal{L}-\delta)^\dagger\tau}$ kommutiert und $(\mathcal{L}-\delta)$ mit $e^{(\mathcal{L}-\delta)\tau}$, ergibt die Produktregel

$$\begin{aligned}
\frac{\mathrm dG}{\mathrm d\tau}
&= \Big[(\mathcal{L}-\delta)^\dagger e^{(\mathcal{L}-\delta)^\dagger\tau}\Big]e^{(\mathcal{L}-\delta)\tau} + e^{(\mathcal{L}-\delta)^\dagger\tau}\Big[e^{(\mathcal{L}-\delta)\tau}(\mathcal{L}-\delta)\Big]
= (\mathcal{L}-\delta)^\dagger G(\tau) + G(\tau)(\mathcal{L}-\delta)
\end{aligned}$$

Integrieren wir das von $0$ bis $\infty$ und ziehen die (konstanten) Matrizen aus dem Integral heraus, so steht links ein Randterm und rechts genau $W$:

$$\underbrace{\Big[G(\tau)\Big]_0^\infty}_{=\;0\;-\;\mathbb{1}\;=\;-\mathbb{1}} = (\mathcal{L}-\delta)^\dagger\underbrace{\int_0^\infty G\,\mathrm d\tau}_{=\;W} + \underbrace{\int_0^\infty G\,\mathrm d\tau}_{=\;W}(\mathcal{L}-\delta) \quad\implies\quad (\mathcal{L}-\delta)^\dagger W + W(\mathcal{L}-\delta) = -\mathbb{1}$$

Dabei ist $G(\infty)=0$ wegen der Hurwitz-Abschätzung aus (a) und $G(0) = \mathbb{1}$.

**(c) Die Lösung ist eindeutig.** Die linke Seite von (7.5) ist eine lineare Abbildung $\mathcal{S}: X\mapsto (\mathcal{L}-\delta)^\dagger X + X(\mathcal{L}-\delta)$ auf dem Raum der Matrizen (ein Sylvester-Operator). Ihre Eigenwerte sind bekanntlich $\overline{\lambda_i}+\lambda_j$, wobei $\lambda_i,\lambda_j$ die Eigenwerte von $\mathcal{L}-\delta$ durchlaufen. Da alle $\mathrm{Re}\,\lambda_i<0$ sind, ist $\mathrm{Re}(\overline{\lambda_i}+\lambda_j)<0$, insbesondere niemals null. $\mathcal{S}$ ist also invertierbar, und die Lösung von $\mathcal{S}(W) = -\mathbb{1}$ ist eindeutig.

**(d) Die Normabschätzung (7.6).** Wir betrachten die Trajektorie $x(\tau) = e^{\mathcal{L}\tau}x_0$ und die „Energie" $V(\tau) := \Vert x(\tau)\Vert_W^2 = x(\tau)^\dagger W x(\tau)$. Mit $\dot x = \mathcal{L}x$ und der Lyapunov-Gleichung (7.5), umgestellt zu $\mathcal{L}^\dagger W + W\mathcal{L} = -\mathbb{1} + 2\delta W$, folgt

$$\begin{aligned}
\frac{\mathrm dV}{\mathrm d\tau}
&= \dot x^\dagger W x + x^\dagger W \dot x
= x^\dagger\big(\mathcal{L}^\dagger W + W\mathcal{L}\big)x
= x^\dagger\big(-\mathbb{1} + 2\delta W\big)x
= \underbrace{-\Vert x\Vert^2}_{\le\,0} + 2\delta\,\underbrace{x^\dagger Wx}_{=\,V}
\;\le\; 2\delta\,V(\tau)
\end{aligned}$$

Das ist eine Grönwall-Ungleichung. Multiplizieren wir mit dem integrierenden Faktor $e^{-2\delta\tau}>0$, so wird die Aussage zu einer Monotonieaussage, die man direkt integrieren kann:

$$\frac{\mathrm d}{\mathrm d\tau}\Big(e^{-2\delta\tau}V(\tau)\Big) = e^{-2\delta\tau}\Big(\underbrace{V'(\tau) - 2\delta V(\tau)}_{\le\,0}\Big) \le 0 \quad\implies\quad e^{-2\delta\tau}V(\tau) \le V(0) \quad\implies\quad \Vert x(\tau)\Vert_W \le e^{\delta\tau}\Vert x_0\Vert_W$$

Da $x_0$ beliebig war, ist das genau $\Vert e^{\mathcal{L}\tau}\Vert_W \le e^{\delta\tau}$. $\;\blacksquare$

Der Kern des Beweises steckt in Schritt (d) in dem Term $-\Vert x\Vert^2$: Die rechte Seite $-\mathbb{1}$ der Lyapunov-Gleichung ist genau die Buchhaltung darüber, dass die Dynamik in der $W$-Metrik in jedem Moment Länge *verliert*. Der Shift $\delta$ erlaubt ihr, diesen Verlust um höchstens $e^{\delta\tau}$ zu unterlaufen.

### Was die Eichung mit dem Propagator macht

> **Proposition 21.** Mit $W$ aus Proposition 20 sei
> $$\tilde P := W^{1/2}\,P\,W^{-1/2},\qquad \tilde{\vec\Sigma}_0 := W^{1/2}\vec{\Sigma}_0 .$$
> Dann gilt
> $$\Vert\tilde P\Vert_2 \le e^{\delta\Delta\tau} \qquad\text{und}\qquad \tilde P^{\,n} = W^{1/2}P^{n}W^{-1/2} , \tag{7.8}$$
> insbesondere ist die Trajektorie $\tilde{\vec\Sigma}_n = \tilde P^{\,n}\tilde{\vec\Sigma}_0$ **exakt** und man liest daraus
> $$\mathrm{vec}\,\rho_S(n\Delta t) = Q\,W^{-1/2}\,\tilde{\vec\Sigma}_n \tag{7.9}$$
> ab. Die Eichung führt also **keinerlei** Näherung ein.

**Beweis.** Für die Norm substituieren wir im Supremum $y = W^{1/2}x$. Weil $W^{1/2}$ invertierbar ist, ist das eine Bijektion, das Supremum läuft also über dieselbe Menge:

$$\begin{aligned}
\Vert\tilde P\Vert_2 = \sup_{y\neq0}\frac{\Vert W^{1/2}PW^{-1/2}y\Vert_2}{\Vert y\Vert_2}
\;\overset{y=W^{1/2}x}{=}\; \sup_{x\neq0}\frac{\Vert W^{1/2}Px\Vert_2}{\Vert W^{1/2}x\Vert_2}
= \sup_{x\neq0}\frac{\Vert Px\Vert_W}{\Vert x\Vert_W}
= \Vert P\Vert_W = \Vert e^{\mathcal{L}\Delta\tau}\Vert_W \;\overset{(7.6)}{\le}\; e^{\delta\Delta\tau}
\end{aligned}$$

Die Potenz ist ein Teleskopprodukt, in dem sich $W^{-1/2}W^{1/2}=\mathbb{1}$ immer wieder aufhebt:

$$\tilde P^{\,n} = \big(W^{1/2}PW^{-1/2}\big)^n = W^{1/2}P\,\underbrace{W^{-1/2}W^{1/2}}_{=\,\mathbb{1}}\,P\,\underbrace{W^{-1/2}W^{1/2}}_{=\,\mathbb{1}}\,P\cdots P\,W^{-1/2} = W^{1/2}P^{n}W^{-1/2}$$

Damit ist $\tilde{\vec\Sigma}_n = \tilde P^n W^{1/2}\vec\Sigma_0 = W^{1/2}P^n\vec\Sigma_0 = W^{1/2}\vec\Sigma_n$, also $\vec\Sigma_n = W^{-1/2}\tilde{\vec\Sigma}_n$, und mit (7.2) und der Projektion $Q$ folgt (7.9). $\;\blacksquare$

Jetzt lesen wir Gleichung (7.3) noch einmal, aber in der neuen Eichung. Das Gatter wird aus $\tilde P$ gebaut, der Skalierungsfaktor ist also $s = \Vert\tilde P\Vert_2 \le e^{\delta\Delta\tau}$, und die Gatter-Strafe über $n$ Schritte wird zu

$$s^{2n} \;\le\; \Big(e^{\delta\Delta\tau}\Big)^{2n} = e^{2\delta\,n\Delta\tau} = e^{2\delta T} \tag{7.10}$$

Das ist der entscheidende Punkt: Die Strafe hängt **nur noch vom Produkt $\delta T$** ab, nicht mehr von der Zahl der Schritte und nicht von der Nicht-Normalität von $P$. Man wählt $\delta$ so klein, dass $\delta T \ll 1$ ist, und Teil 1 aus (7.3) ist erledigt. Teil 2 wird jetzt ebenfalls in der $W$-Norm gemessen, und dort ist es nach Proposition 20 **automatisch** durch $e^{2\delta T}$ nach oben beschränkt — das Wachstum auf $1{,}21\times10^{6}$ kann gar nicht mehr auftreten. Gemessen bei $\delta = 0{,}1$ fällt $\Vert\tilde P^{\,n}\tilde{\vec\Sigma}_0\Vert/\Vert\tilde{\vec\Sigma}_0\Vert$ über 100 Schritte auf $0{,}455$, Teil 2 ist also $0{,}21$. Genau das ist der Punkt der Eichung: Sie macht **beide** Faktoren einzeln zu Zahlen der Ordnung $1$, statt zwei astronomische Zahlen gegeneinander wegkürzen zu lassen.

**Warum darf $\delta$ nicht null sein?** Weil $\mathcal{L}$ den Eigenwert $0$ besitzt. Das ist keine Panne, sondern Physik: Der thermische Gleichgewichtszustand $\vec\Sigma_{\mathrm{ss}}$ ist stationär, erfüllt also $\mathcal{L}\vec\Sigma_{\mathrm{ss}} = 0$. Entlang dieser Richtung ist $\Vert e^{(\mathcal{L}-\delta)\tau}\vec\Sigma_{\mathrm{ss}}\Vert^2 = e^{-2\delta\tau}\Vert\vec\Sigma_{\mathrm{ss}}\Vert^2$, und das Integral (7.7) liefert dort den Beitrag $\Vert\vec\Sigma_{\mathrm{ss}}\Vert^2/(2\delta)$ — endlich für jedes $\delta>0$, divergent für $\delta=0$. Der Shift ist also exakt so groß zu wählen, dass er dieses eine marginale Mode zähmt, und nicht größer.

**Was kostet $\delta$?** Genau den Faktor $e^{-2\delta T}$ aus (7.10) und sonst nichts. Insbesondere wird er *nicht* gegen die Kondition von $W$ eingetauscht, wie man befürchten könnte. Gemessen für unser Modell ($T = 1000\,\mathrm{fs}$, also $T = 0{,}1884\,\mathrm{cm}$):

## 7.4 Jetzt wird die Arnoldi-Kompression gratis

$\tilde P$ lebt immer noch auf $\mathbb{C}^{2640}$, das wären $\lceil\log_2 2640\rceil + 1 = 13$ Qubits und ein $8192\times8192$-Gatter. Wir komprimieren also wie im Abschnitt *Arnoldi compression*, nur jetzt auf dem geeichten Propagator: Sei $\tilde{\mathcal{K}}_m := \mathcal{K}_m(\tilde P,\tilde{\vec\Sigma}_0)$ der Krylov-Raum, $\tilde Q_m$ seine Arnoldi-Basis und $\tilde H_m := \tilde Q_m^\dagger \tilde P\,\tilde Q_m$.

> **Proposition 22.** $\Vert \tilde H_m\Vert_2 \le \Vert\tilde P\Vert_2 \le e^{\delta\Delta\tau}$ für **jedes** $m$.

**Beweis.** Das ist Proposition 16, wörtlich, mit $E\to\tilde P$ — der dortige Beweis benutzte nur, dass $\tilde Q_m$ eine Isometrie ist. Neu ist allein die zweite Ungleichung, und die ist Proposition 21. $\;\blacksquare$

So harmlos diese Zeile aussieht, sie räumt den unangenehmsten Punkt der bisherigen Kompression weg. Dort war die Wahl von $m$ ein echter Trade-off: Ein zu großes $m$ zwang das Arnoldi-Verfahren, über die physikalisch besuchte Trajektorie hinauszugreifen, nahm dabei die normaufblähenden Richtungen aus Proposition 14a in die Basis auf, ließ $\Vert H_m\Vert_2$ wieder gegen $\sqrt2$ steigen und die Erfolgswahrscheinlichkeit einbrechen (beim Pfadintegral schon bei $m=64$: $\Vert H_m\Vert_2 = 1{,}243$ und $p = 2{,}5\times10^{-18}$).

Der Grund dafür war, dass die Schranke $\Vert H_m\Vert \le \Vert E\Vert = \sqrt2$ zwar galt, aber nutzlos war — sie erlaubte genau die Katastrophe. In der neuen Eichung lautet dieselbe Schranke $\Vert\tilde H_m\Vert \le 1{,}000038$, und *die* erlaubt nichts mehr. Egal wie groß $m$ wird, egal welche Richtungen die Basis noch aufnimmt: Die komprimierte Matrix bleibt eine Kontraktion. **$m$ ist damit kein Kompromiss mehr, sondern ein reiner Konvergenzparameter**, den man so lange erhöht, bis die Genauigkeit stimmt — zum Preis von $\lceil\log_2 m\rceil$ Qubits und sonst nichts.

Dass die ersten $m$ Schritte dabei sogar exakt herauskommen, ist eine Standardeigenschaft des Verfahrens, die man in unserer Notation schnell nachrechnet:

> **Proposition 23.** Für $k \le m-1$ gilt
> $$\tilde P^{\,k}\,\tilde{\vec\Sigma}_0 = \Vert\tilde{\vec\Sigma}_0\Vert\;\tilde Q_m\,\tilde H_m^{\,k}\,e_1 , \tag{7.11}$$
> die Kompression reproduziert die ersten $m$ Zeitschritte also **ohne jeden Fehler**, auch wenn $h_{m+1,m}\neq0$ ist.

**Beweis.** Wir brauchen zuerst eine kleine Beobachtung über Hessenberg-Matrizen.

*Hilfssatz:* Ist $H$ obere Hessenberg-Matrix, so hat $H^k e_1$ höchstens die Einträge $1,\dots,k+1$ von null verschieden. — Denn Spalte $i$ von $H$ ist nach Definition auf die Zeilen $1,\dots,i+1$ beschränkt; ist also ein Vektor $v$ auf die Indizes $1,\dots,j$ beschränkt, so ist $Hv$ eine Linearkombination der Spalten $1,\dots,j$ und damit auf $1,\dots,j+1$ beschränkt. Mit $e_1$ (beschränkt auf Index $1$) folgt die Behauptung per Induktion.

Nun die eigentliche Aussage, ebenfalls per Induktion über $k$. Für $k=0$ ist mit $q_1 = \tilde{\vec\Sigma}_0/\Vert\tilde{\vec\Sigma}_0\Vert$ trivialerweise $\tilde Q_m e_1 = q_1$, also stimmt (7.11). Gelte (7.11) für ein $k\le m-2$. Wir wenden $\tilde P$ an und setzen die Arnoldi-Relation (6.3) ein:

$$\begin{aligned}
\tilde P^{\,k+1}\tilde{\vec\Sigma}_0
&= \Vert\tilde{\vec\Sigma}_0\Vert\;\tilde P\,\tilde Q_m\,\tilde H_m^{\,k}e_1
= \Vert\tilde{\vec\Sigma}_0\Vert\;\Big(\tilde Q_m\tilde H_m + h_{m+1,m}\,q_{m+1}e_m^{\mathsf T}\Big)\tilde H_m^{\,k}e_1 \\[2pt]
&= \Vert\tilde{\vec\Sigma}_0\Vert\;\tilde Q_m\tilde H_m^{\,k+1}e_1 \;+\; \Vert\tilde{\vec\Sigma}_0\Vert\;h_{m+1,m}\,q_{m+1}\underbrace{\Big(e_m^{\mathsf T}\,\tilde H_m^{\,k}e_1\Big)}_{=\;0\;\text{für}\;k\le m-2} = \Vert\tilde{\vec\Sigma}_0\Vert\;\tilde Q_m\tilde H_m^{\,k+1}e_1
\end{aligned}$$

Der Störterm verschwindet, weil $e_m^{\mathsf T}\tilde H_m^{\,k}e_1$ gerade der $m$-te Eintrag von $\tilde H_m^{\,k}e_1$ ist, dieser Vektor nach dem Hilfssatz aber nur bis zum Index $k+1 \le m-1$ besetzt ist. $\;\blacksquare$

**Gemessen.** Alle Werte über 100 Zeitschritte gegen den `HEOMSolver` von QuTiP, bei $\delta = 0{,}02\,\mathrm{cm}^{-1}$. Die dritte Spalte zeigt $\Vert\tilde H_m\Vert_2 = 1{,}000038$ — für jedes einzelne $m$ exakt derselbe Wert, genau wie Proposition 22 es verlangt:

| ---: | ---: | ---: | ---: | ---: |
| $16$ | $5$ | $1{,}000038$ | $1{,}93\times10^{-1}$ | $0{,}172$ |
| $32$ | $6$ | $1{,}000038$ | $6{,}22\times10^{-2}$ | $0{,}332$ |
| $64$ | $7$ | $1{,}000038$ | $\mathbf{5{,}27\times10^{-7}}$ | $0{,}365$ |
| $96$ | $8$ | $1{,}000038$ | $5{,}27\times10^{-7}$ | $0{,}365$ |

Zwei Dinge sind wichtig. Erstens läuft die Genauigkeit bei $5{,}27\times10^{-7}$ in einen Boden — und das ist nicht *unser* Fehler: Iteriert man den vollen $2640\times2640$-Propagator $P$ ohne jede Kompression und ohne Eichung, kommt exakt dieselbe Zahl heraus. Es ist die Integrationstoleranz des Referenzsolvers. Um zu sehen, was die Kompression wirklich kostet, muss man deshalb gegen die exakte geeichte Trajektorie vergleichen statt gegen QuTiP; über 500 Schritte (gemessen bei $\delta = 0{,}1$) fällt dieser reine Kompressionsfehler von $3{,}15\times10^{-1}$ bei $m=32$ über $2{,}91\times10^{-2}$ ($m=64$) und $1{,}97\times10^{-4}$ ($m=96$) auf $1{,}65\times10^{-8}$ bei $m=128$ und $3{,}35\times10^{-14}$ bei $m=192$ — monoton, ohne Umkehrpunkt.

Zweitens bleibt die letzte Spalte konstant, sobald $m$ groß genug ist. Genau diese Eigenschaft fehlte ohne Eichung: dort wäre $p_{\mathrm{total}}$ ab einem gewissen $m$ wieder eingebrochen.

**Registergröße.** Von $\mathcal{D}_{\mathrm{tot}} = 2640$ (12 System-Qubits) schrumpfen wir auf $m = 128$ (7 System-Qubits), zusammen mit der Ancilla also **8 Qubits**. Bei $m=64$ liegt Route C mit 7 Qubits schon bei $5{,}27\times10^{-7}$.

## 7.5 Das Gitter und die Ablesung

Die $m\times m$-Matrix $\tilde H_m$ wird auf $m_p = 2^{\lceil\log_2 m\rceil}$ aufgefüllt und **einmal** nach Kapitel 5 dilatiert:

$$U = \begin{pmatrix} \tilde H_m/s & B\\ C & -(\tilde H_m/s)^\dagger\end{pmatrix},\qquad B = \sqrt{\mathbb{1}-\tfrac{\tilde H_m\tilde H_m^\dagger}{s^2}},\qquad C = \sqrt{\mathbb{1}-\tfrac{\tilde H_m^\dagger\tilde H_m}{s^2}},\qquad s = \Vert\tilde H_m\Vert_2$$

Danach besteht der gesamte Schaltkreis aus genau einem Gatter, das man wiederholt:

```
krylov (7 Qubits) : ────[ U ]─────────────[ U ]─────────────[ U ]────────── ... ───> Y_n
                         │                 │                 │
Ancilla           : |0>──[ U ]─(M)──|0>────[ U ]─(M)──|0>────[ U ]─(M)─|0>─ ...
                               │                 │                 │
klass. Register   : ───────────●─────────────────●─────────────────●─────── ...
                             rec[1]            rec[2]            rec[3]
```

Zwei Unterschiede zum Bild in dem Abschnitt *Die Umsetzung auf dem Quantencomputer* fallen auf.

**Erstens brauchen wir keine Zustandspräparation.** Der Startvektor des Krylov-Raums ist per Konstruktion $q_1 = \tilde{\vec\Sigma}_0/\Vert\tilde{\vec\Sigma}_0\Vert$, in Krylov-Koordinaten also $y_0 = \Vert\tilde{\vec\Sigma}_0\Vert\,e_1$. Das Register startet damit ohnehin schon im richtigen Zustand $\vert 0\cdots0\rangle$; die gesamte Information über $\rho_S(0)$ und die Eichung steckt in der klassischen Zahl $\Vert\tilde{\vec\Sigma}_0\Vert$ und in der Basis $\tilde Q_m$.

**Zweitens messen wir die Ancilla nach jedem Schritt und setzen sie zurück**, statt für jeden Schritt ein frisches Ancilla-Qubit zu spendieren. Nach dem Prinzip der aufgeschobenen Messung, das wir im Abschnitt *Many steps* diskutiert haben, ist das exakt äquivalent: Sobald Schritt $t$ vorbei ist, berührt kein Gatter dieses Ancilla-Qubit je wieder, die Messung kommutiert also mit allem Folgenden. Der Vorteil ist, dass wir mit **einem** Ancilla-Qubit für beliebig viele Schritte auskommen, statt $n$ davon zu brauchen. Das Messprotokoll wandert stattdessen in das klassische Register `rec`.

### Der akzeptierte Zweig ist deterministisch

> **Proposition 24.** Bedingt auf das Messprotokoll $\mathtt{rec} = 0\cdots0$ ist der Zustand des Krylov-Registers nach $t$ Schritten
> $$\vert Y_t\rangle = \frac{\tilde H_m^{\,t}\,y_0}{\Vert\tilde H_m^{\,t}\,y_0\Vert} \tag{7.12}$$
> und damit **unabhängig von jedem Zufall in den Messungen**.

**Beweis.** Induktion über $t$. Für $t=0$ ist nichts zu zeigen. Sei das Register vor Schritt $t$ im reinen Produktzustand $\vert Y_{t-1}\rangle\otimes\vert0\rangle_E$. Die Anwendung von $U$ spaltet ihn nach dem Abschnitt *One step* in genau zwei Zweige auf:

$$U\big(\vert0\rangle_E\otimes\vert Y_{t-1}\rangle\big) = \vert0\rangle_E\otimes\Big(\tfrac{\tilde H_m}{s}Y_{t-1}\Big) \;+\; \vert1\rangle_E\otimes\big(C\,Y_{t-1}\big)$$

Die Messung wählt zufällig einen der beiden Zweige aus — welchen, das ist die einzige Zufälligkeit im ganzen Ablauf. *Gegeben* das Ergebnis $\vert0\rangle$ projiziert der Kollaps aber auf den ersten Summanden, und die Normierung, die die Natur automatisch vornimmt, hebt den Faktor $1/s$ wieder auf:

$$\vert Y_t\rangle = \frac{\big(\tilde H_m/s\big)Y_{t-1}}{\big\Vert\big(\tilde H_m/s\big)Y_{t-1}\big\Vert} = \frac{\tfrac1s\,\tilde H_mY_{t-1}}{\tfrac1s\,\Vert\tilde H_mY_{t-1}\Vert} = \frac{\tilde H_m\,Y_{t-1}}{\Vert\tilde H_m\,Y_{t-1}\Vert}$$

Das ist eine **Funktion von $Y_{t-1}$ allein**, ohne jeden Rest an Zufall. Das anschließende `reset` bringt die Ancilla wieder nach $\vert0\rangle$, so dass die Induktionsvoraussetzung für Schritt $t+1$ wiederhergestellt ist. Setzt man die Rekursion $t$-mal ein und benutzt $y_0 = \Vert\tilde{\vec\Sigma}_0\Vert\,e_1 \propto Y_0$, folgt (7.12). $\;\blacksquare$

Das hat eine sehr praktische Konsequenz. Der Zufall im Schaltkreis steckt *ausschließlich* in der Frage, **ob** ein Durchlauf akzeptiert wird, nicht darin, **was** man im Erfolgsfall vorfindet. Ein einziger akzeptierter Shot trägt daher bereits den exakten Zustand; die Statistik über viele Shots wird nur für die skalare Zahl $p_{\mathrm{total}}$ gebraucht.

> **Praktischer Hinweis zur Simulation.** Qiskit Aer implementiert eine Messung mitten im Schaltkreis, indem es ein Ergebnis **würfelt** — und zwar in der `density_matrix`-Methode genauso wie in der `statevector`-Methode. Ein Zwei-Qubit-Test mit exakter $0{,}3/0{,}7$-Aufspaltung liefert bei einem Shot $[0,\,1]$ und bei 1024 Shots $[0{,}284,\,0{,}716]$. Wer den akzeptierten Zweig also aus einer gespeicherten Dichtematrix ausliest (etwa indem er die Ancilla misst und die Fehlzweige per `if_test` in einen Mülleimer-Zustand schiebt), rechnet in Wahrheit Monte-Carlo und handelt sich einen Fehler der Ordnung $1/\sqrt{\mathrm{shots}}$ ein. Proposition 24 sagt, dass das unnötig ist: `save_statevector(conditional=True)` gruppiert die gespeicherten Zustände nach dem klassischen Protokoll, und der Eintrag zum Schlüssel $0\cdots0$ **ist** der akzeptierte Zweig — exakt und ohne Sampling.

### Die Rückrechnung auf die Physik

Der Quantencomputer gibt uns den normierten Vektor $Y_t$. Um daraus $\rho_S(t)$ zu machen, brauchen wir zwei Dinge: seine Länge und die Rücktransformation aus den Krylov- und Eichkoordinaten. Die Länge liefert Proposition 13, wörtlich übertragen mit $E\to\tilde H_m$:

$$p_{\mathrm{total}}(t) = \frac{\Vert\tilde H_m^{\,t}y_0\Vert^2}{s^{2t}\Vert y_0\Vert^2} \quad\Longrightarrow\quad \Vert y_t\Vert \;=\; \underbrace{\Vert\tilde{\vec\Sigma}_0\Vert}_{=\,\Vert y_0\Vert}\cdot\, s^{\,t}\,\sqrt{p_{\mathrm{total}}(t)} \tag{7.13}$$

Die Rücktransformation fassen wir in einer einzigen klassischen Matrix zusammen, die **einmal** vorab berechnet wird:

$$R := \Big(W^{-1/2}\,\tilde Q_m\Big)_{[\,1:d^2,\;:\,]} \;\in\;\mathbb{C}^{d^2\times m} \qquad\Longrightarrow\qquad \mathrm{vec}\,\rho_S(t\Delta t) = R\;y_t \tag{7.14}$$

Denn nach (7.9) ist $\mathrm{vec}\rho_S = QW^{-1/2}\tilde{\vec\Sigma}_t$, nach der Kompression ist $\tilde{\vec\Sigma}_t \approx \tilde Q_m y_t$, und $Q$ schneidet genau die ersten $d^2$ Zeilen heraus. Zusammengesetzt:

$$\mathrm{vec}\,\rho_S(t\Delta t) = \underbrace{Q\,W^{-1/2}\tilde Q_m}_{=\;R}\;\underbrace{\Vert\tilde{\vec\Sigma}_0\Vert\,s^{\,t}\sqrt{p_{\mathrm{total}}(t)}}_{=\;\lambda_t}\;\cdot\;Y_t$$

| Größe | Bedeutung | Woher kommt der Wert? | Wann im Ablauf? |
| :--- | :--- | :--- | :--- |
| $\Vert\tilde{\vec\Sigma}_0\Vert$ | Länge des geeichten Startvektors | $\Vert W^{1/2}\vec\Sigma_0\Vert$ | **vor** dem Quanten-Run, rein klassisch |
| $s$ | Dilatations-Skalierung | $s = \Vert\tilde H_m\Vert_2 \le e^{\delta\Delta\tau}$ | **vor** dem Quanten-Run, rein klassisch |
| $R$ | Rücktransformation in den Systemblock | $R = (W^{-1/2}\tilde Q_m)_{[1:d^2,:]}$ | **vor** dem Quanten-Run, rein klassisch |
| $p_{\mathrm{total}}(t)$ | kumulierte Erfolgswahrscheinlichkeit | Anteil der Shots mit $\mathtt{rec}[1..t] = 0\cdots0$ | **nach** dem Run, aus dem klassischen Register |
| $Y_t$ | normierter Registerzustand | Zustand der $\lceil\log_2 m\rceil$ Krylov-Qubits, bedingt auf die akzeptierten Shots | am Zeitschritt $t$ |

**Eine zweite, unabhängige Ablesung.** Es gibt einen Weg, der ohne $p_{\mathrm{total}}$ auskommt. Die HEOM-Dynamik erhält die Spur des nullten ADOs, es gilt also zu jeder Zeit exakt $\mathrm{Tr}\,\rho_S(t) = 1$. Das ist keine Zusatzinformation über die Lösung, sondern eine Eigenschaft der Bewegungsgleichung, und sie legt den fehlenden Skalar fest:

$$\rho_S(t) = \frac{\mathrm{unvec}\big(R\,Y_t\big)}{\mathrm{Tr}\,\mathrm{unvec}\big(R\,Y_t\big)}$$

Nebenbei erledigt das auch die globale Phase, die ein normierter Quantenzustand ohnehin nur bis auf $e^{i\varphi}$ festlegt: Die physikalische Spur muss reell und positiv sein. Beide Ablesungen benutzen **disjunkte** Information — die eine das Messprotokoll, die andere die Spurerhaltung —, ihre Übereinstimmung ist also ein in sich geschlossener Test des ganzen Kapitels.

zu lösen (Hammarling-Verfahren), statt $W$ zu bilden — hier nicht umgesetzt.

## 7.7 Wo die Arbeit anfällt — und was das Verfahren nicht leistet

**Die Aufteilung.** Alles Klassische wird genau einmal gerechnet, und keine dieser Rechnungen wächst mit der Zahl der Zeitschritte oder mit der Endzeit $T$:

| Schritt | Aufwand | wächst mit $n$? |
| :--- | :--- | :---: |
| $\mathcal{L}_{\mathrm{HEOM}}$ aufstellen | dünn besetzt, aus QuTiP | nein |
| Lyapunov-Gleichung (7.5) lösen | $\mathcal{O}(\mathcal{D}_{\mathrm{tot}}^3)$, $\approx90\,\mathrm{s}$ bei $2640$ | nein |
| $P = e^{\mathcal{L}\Delta\tau}$ | $\mathcal{O}(\mathcal{D}_{\mathrm{tot}}^3)$, $\approx37\,\mathrm{s}$ | nein |
| Arnoldi (6.3) auf $\tilde P$ | $m$ Matrix-Vektor-Produkte | nein |
| Dilatation (Sz.-Nagy) | $\mathcal{O}(m^3)$, $m\le128$ | nein |
| **Propagation** | $n$ Anwendungen **eines** Gatters auf $\lceil\log_2m\rceil+1$ Qubits | **auf dem Gitter** |

Klassisch propagierte Zeitschritte der *Trajektorie* $\rho_S(t)$: **null**, für jedes $T$. Route B braucht $K$ davon, bevor das Gitter überhaupt starten darf, eine vollständig klassische Lösung braucht alle $n$.

**Diese Formulierung ist allerdings zu günstig, und das muss man sehen.** Der Krylov-Raum ist per Definition

$$\mathcal K_m(\tilde P,\tilde{\vec\Sigma}_0) = \mathrm{span}\{\tilde{\vec\Sigma}_0,\;\tilde P\tilde{\vec\Sigma}_0,\;\dots,\;\tilde P^{\,m-1}\tilde{\vec\Sigma}_0\},$$

er *enthält* also gerade die ersten $m$ Zeitschritte, und ihn aufzubauen kostet $m$ Matrix-Vektor-Produkte. Bei $m=128$ und $n=100$ hat der klassische Computer damit rechnerisch mehr getan, als das Gitter anschließend leistet — strukturell dasselbe Problem wie das Fenster $K$ bei Route B, nur an anderer Stelle. Der Unterschied, und er ist real: **$m$ hängt nicht von $n$ ab.** Dasselbe $m=128$ trägt 100, 500 und 1000 Schritte (Abschnitt 7.6), während $K$ mitwachsen müsste, um die Genauigkeit zu halten. Der Gewinn beginnt also erst bei $n\gg m$, und dort ist er dann unbegrenzt.

**Woran die Genauigkeit wirklich gemessen ist.** Alle Zahlen in Abschnitt 7.6
lauten „Abweichung von `qutip.HEOMSolver` bei Tiefe 3 und $N_k=1$". Das ist nicht
dasselbe wie „Abweichung von der Physik". Gemessen (reiner QuTiP gegen reinen
QuTiP):

$$\big|\text{Tiefe }4 - \text{Tiefe }3\big| = 2{,}3\times10^{-3},\qquad
\big|\text{Tiefe }2 - \text{Tiefe }3\big| = 1{,}3\times10^{-2},\qquad
\big|N_k{=}2 - N_k{=}1\big| = 1{,}2\times10^{-5}$$

Die Matsubara-Zahl ist also konvergiert, die Hierarchietiefe nicht: Tiefe 3 ist
selbst nur auf rund $10^{-3}$ genau. Unsere $5\times10^{-7}$ sind damit der
Abstand zur *selben* Näherung — die eigentliche Schranke gegenüber der wahren
Dynamik des Modells ist rund vier Größenordnungen größer und wird von der
HEOM-Abschneidung gesetzt, die wir mit dem Referenzsolver teilen. Für die
Zielvorgabe $10^{-4}$ heißt das: gegenüber dem gewählten HEOM-Modell wird sie um
drei Größenordnungen übertroffen, gegenüber der exakten Physik bei Tiefe 3 nicht.

**Was das nicht heißt.** Die Vorbereitung ist nicht umsonst. Die Lyapunov-Gleichung und das Matrixexponential sind $\mathcal{O}(\mathcal{D}_{\mathrm{tot}}^3)$-Operationen auf dem vollen ADO-Raum, und für ein Vier-Site-Modell ist das durchaus vergleichbar damit, die Dynamik gleich klassisch auszurechnen. Die Aussage dieses Kapitels ist eine über **Skalierung**, nicht über einen Geschwindigkeitsvorteil gegenüber QuTiP bei 4 Sites: Der klassische Aufwand ist hier $\mathcal{O}(1)$ in der Zahl der Zeitschritte, während Route B $K$ Propagationsschritte bezahlt und die rein klassische Lösung $n$.

**Grenzen der Systemgröße.** Der Speicher wird von $\mathcal{D}_{\mathrm{tot}} = \mathcal{N}_{\mathrm{ADOs}}\cdot d^2$ diktiert, und die Lyapunov-Lösung braucht mehrere $\mathcal{D}_{\mathrm{tot}}\times\mathcal{D}_{\mathrm{tot}}$-Matrizen gleichzeitig:

* $N=4$, Tiefe 3: $165$ ADOs, $\mathcal{D}_{\mathrm{tot}} = 2640$, rund $110\,\mathrm{MB}$ pro Matrix — bequem.
* $N=7$, Tiefe 2: $120$ ADOs, $\mathcal{D}_{\mathrm{tot}} = 5880$, rund $550\,\mathrm{MB}$ pro Matrix und etwa eine halbe Stunde — machbar.
* $N=7$, Tiefe 3: $680$ ADOs, $\mathcal{D}_{\mathrm{tot}} = 33320$, rund $18\,\mathrm{GB}$ pro Matrix — außer Reichweite.

**Zwei offene Enden.** Beide betreffen die Vorbereitung, nicht das Verfahren selbst:

1. *Die Eichung billiger bekommen.* $W$ muss nicht aus einer Lyapunov-Gleichung kommen — jedes $W\succ0$, das (7.6) erfüllt, tut es. Naheliegend wäre eine physikalisch motivierte diagonale Umskalierung der ADOs, wie sie in der HEOM-Literatur ohnehin üblich ist; die wäre praktisch gratis. Ein rein diagonales $W$ aus `scipy.linalg.matrix_balance` schafft allerdings nur $\Vert P\Vert:32{,}4\to2{,}54$, ist also für sich genommen nicht genug — es müsste eine klügere Diagonale sein.
2. *Das Matrixexponential vermeiden.* Statt $P = e^{\mathcal{L}\Delta\tau}$ dicht auszurechnen, könnte man direkt auf dem geeichten Erzeuger $\tilde{\mathcal{L}} = W^{1/2}\mathcal{L}W^{-1/2}$ Arnoldi laufen lassen und erst die kleine $m\times m$-Matrix exponenzieren. Aus (7.5) folgt nämlich $\tilde{\mathcal{L}}+\tilde{\mathcal{L}}^\dagger = -W^{-1/2}W^{-1/2} + 2\delta\mathbb{1} \preceq 2\delta\mathbb{1}$, und diese Eigenschaft überträgt sich wie in Proposition 22 auf jede Kompression $\tilde Q_m^\dagger\tilde{\mathcal{L}}\tilde Q_m$ — die Kontraktivität bliebe also erhalten. Hier ist das nicht umgesetzt.

**Und die Ablesung selbst?** $\mathrm{vec}\,\rho_S = R\,y_t$ ist ein *lineares* Funktional des Registerzustands und damit keine Observable, die man direkt messen könnte. Auf echter Hardware braucht man dafür Tomographie des kleinen Krylov-Registers oder einen Hadamard-Test gegen einen Referenzzustand — das ist aber exakt dieselbe Situation wie bei Route B (siehe die Tabelle in dem Abschnitt *Die Umsetzung auf dem Quantencomputer*) und wird durch die Eichung weder besser noch schlechter. Was die Eichung ändert, ist ausschließlich, dass es überhaupt akzeptierte Shots gibt, an denen man diese Tomographie durchführen könnte: $0{,}37$ statt $10^{-301}$.

## Zusammenfassung des Kapitels

Der Weg in einem Absatz: Die HEOM-Hierarchie ist auf dem erweiterten Raum bereits eine Halbgruppe, $\vec\Sigma_n = P^n\vec\Sigma_0$ ist exakt (Proposition 18), und der ganze Gedächtnis-Rest $R_n$ aus Route B entsteht erst dadurch, dass man mit $Q$ auf den Systemblock projiziert und die ADOs wegwirft. Wer sie behält, braucht kein $K$ und keine klassisch vorberechnete Historie. Das einzige Hindernis ist dann, dass $P$ mit $\rho(P)=1$ zwar brav, mit $\Vert P\Vert_2 = 32{,}39$ aber stark nicht-normal ist, und dass die Sz.-Nagy-Dilatation stur die Operatornorm berechnet — ein Hindernis, das auch feinere Zeitschritte nicht beseitigen (Proposition 19). Die Auflösung ist, dass $32{,}39$ eine Aussage über das *Skalarprodukt* ist: Die Lyapunov-Gleichung (7.5) liefert eine Metrik $W$, in der $P$ eine Kontraktion bis auf $e^{\delta\Delta\tau}$ ist (Proposition 20), ohne die Trajektorie im Geringsten zu verändern (Proposition 21). In dieser Eichung wird die Arnoldi-Kompression zum reinen Konvergenzparameter (Propositionen 22 und 23), und der akzeptierte Zweig des Schaltkreises ist deterministisch (Proposition 24).

$$\boxed{\;\Vert P\Vert_2 = 32{,}39 \;\xrightarrow{\;W^{1/2}\cdot W^{-1/2}\;}\; \Vert\tilde P\Vert_2 = 1{,}000038 \qquad\Longrightarrow\qquad p_{\mathrm{total}}(100):\;\;10^{-301}\;\longrightarrow\;0{,}37\;}$$

| Proposition | Aussage |
| :---: | :--- |
| **18** | HEOM ist auf dem ADO-Raum eine Halbgruppe; $R_n \equiv 0$ ohne jedes $K$ |
| **19** | $n\ln s \to T\,\omega(\mathcal{L})$: feinere Zeitschritte helfen nicht |
| **20** | Lyapunov-Eichung: $\exists\,W\succ0$ mit $\Vert e^{\mathcal{L}\tau}\Vert_W \le e^{\delta\tau}$ |
| **21** | $\tilde P = W^{1/2}PW^{-1/2}$ ist ähnlich zu $P$ — die Eichung ist exakt |
| **22** | $\Vert\tilde H_m\Vert_2 \le e^{\delta\Delta\tau}$ für jedes $m$ — Kompression bleibt Kontraktion |
| **23** | die ersten $m$ Zeitschritte sind exakt, auch bei $h_{m+1,m}\neq0$ |
| **24** | der auf $0\cdots0$ bedingte Zweig ist deterministisch — ein Shot genügt für den Zustand |

Die Implementierung liegt in `heom_gauge.py` (Erzeuger, Eichung, Arnoldi, Dilatation) und `gauge_circuit.py` (das Gitter); die Rechnungen und alle Abbildungen stehen in `main_durchgaengiges_grid.ipynb`.
